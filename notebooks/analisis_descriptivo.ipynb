# Análisis del Dataset SIVIGILA - VIH/SIDA

Este notebook explora el dataset de vigilancia epidemiológica para encontrar patrones sobre los casos registrados.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar dataset
df = pd.read_csv('../data/sivigila_vih(in).csv', encoding='latin1', low_memory=False)
df.head()
```

## Limpieza de datos
Filtramos las columnas de interés y revisamos valores nulos.


```python
# Seleccionar columnas relevantes para el análisis
df_clean = df[['fec_not', 'edad_', 'sexo_', 'nombre_barrio', 'grupo_edad_quinque', 'nom_eve', 'tip_ss_']].copy()

# Ver nulos
print(df_clean.isnull().sum())

# Eliminar nulos en sexo y edad
df_clean = df_clean.dropna(subset=['sexo_', 'edad_'])
```

## Análisis exploratorio y Visualización


```python
# Casos por género
plt.figure(figsize=(8,5))
sns.countplot(data=df_clean, x='sexo_')
plt.title('Distribución de Casos por Sexo')
plt.show()
```

```python
# Distribución de edad de los casos
plt.figure(figsize=(10,5))
# Filtrar edades irreales si las hay
df_age = df_clean[(df_clean['edad_'] >= 0) & (df_clean['edad_'] < 120)]
sns.histplot(data=df_age, x='edad_', bins=30, kde=True)
plt.title('Distribución de Edad de los Casos')
plt.show()
```

```python
# Casos por tipo de seguridad social (tip_ss_)
plt.figure(figsize=(8,5))
sns.countplot(data=df_clean, x='tip_ss_')
plt.title('Distribución de Casos por Tipo de Seguridad Social')
plt.show()
```

## Conclusiones
1. **Género:** La distribución de casos muestra las proporciones entre hombres (M) y mujeres (F).
2. **Edad:** Se puede observar en qué rango de edad se concentran más los casos reportados (usualmente adultos jóvenes y de mediana edad).
3. **Seguridad Social:** Permite identificar a qué régimen (Subsidiado, Contributivo, etc.) pertenecen la mayoría de los casos.
